```python
# ================================
# Import Libraries
# ================================
import pandas as pd
import numpy as np

# ================================
# Number of rows (1 Million)
# ================================
n = 1000000

# ================================
# Generate Dataset
# ================================
df = pd.DataFrame({
    "order_id": np.arange(1, n+1),
    "customer_id": np.random.randint(10000, 99999, n),

    "product": np.random.choice(
        ["Shoes", "Mobile", "Laptop", "Watch", "Bag", "Headphones", "TV"], n
    ),

    "category": np.random.choice(
        ["Electronics", "Fashion", "Home", "Sports"], n
    ),

    "price": np.random.randint(500, 100000, n),
    "quantity": np.random.randint(1, 5, n),

    "discount_percent": np.random.randint(0, 70, n),

    "payment_mode": np.random.choice(
        ["UPI", "Card", "COD", "NetBanking"], n
    ),

    "city": np.random.choice(
        ["Delhi", "Mumbai", "Lucknow", "Kanpur", "Noida", "Pune"], n
    ),

    "rating": np.round(np.random.uniform(1, 5, n), 1)
})

# ================================
# Create Extra Columns
# ================================
df["discounted_price"] = df["price"] - (df["price"] * df["discount_percent"] / 100)
df["total_amount"] = df["discounted_price"] * df["quantity"]

# ================================
# Add Date Column
# ================================
df["order_date"] = pd.date_range(start="2023-01-01", periods=n, freq="min")

# ================================
# Fraud Logic
# ================================
df["fraud_flag"] = 0

df.loc[(df["price"] > 70000) & (df["quantity"] >= 3), "fraud_flag"] = 1
df.loc[df["discount_percent"] > 60, "fraud_flag"] = 1
df.loc[(df["rating"] < 2) & (df["total_amount"] > 150000), "fraud_flag"] = 1

# ================================
# Save Dataset
# ================================
df.to_csv("fraud_ecommerce_1M.csv", index=False)

print("✅ 1 Million Fraud Dataset Generated Successfully")
```
